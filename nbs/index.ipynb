# house_pricing_nbdev

> A reusable nbdev library for training, evaluating, saving, and scoring a housing price machine learning pipeline.


## What it does

`house_pricing_nbdev` software converts the ames housing workflow in into a package API. It includes the selected feature set, feature-engine
preprocessing, custom sklearn-compatible transformers, Lasso price modeling, evaluation helpers, Kaggle-style submission creation, and
model persistence.


## Installation

Install in development mode from the repository root:

```sh
pip install -e ".[dev]"
```

Then export and test notebook changes with:

```sh
nbdev-export
nbdev-test
nbdev-prepare
```


## Quick start

Load the included training data from `script/train.csv`:


````python
train = load_housing_data()
train.shape
````

Build the same end-to-end pipeline used by the original source notebook:


````python
pipeline = build_price_pipeline()
pipeline.steps[-1][0]
````


